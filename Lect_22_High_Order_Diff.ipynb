Still more accurate approximations to $f'(x_0)$ can be obtained by taking linear 
combinations of more Taylor Theorem approximations and solving for $f'(x_0)$.  
For example, Taylor Theorem approximations for each of

$$
f(x_0-2h),f(x_0-h),f(x_0+2h),f(x_0+h)
$$

can be used to derive an $O(h^4)$ approximation. Consider the following four 
applications of Taylor's theorem,

\begin{align}
&f(x_0-2h)=f(x_0)-2hf'(x_0)+2h^2f''(x_0)-\frac{8h^3}{6}f'''(x_0)+\frac{16h^4}{24}f^{(4)}(x_0)+O(h^5) \tag{1}\\
&f(x_0-h)=f(x_0)-hf'(x_0)+\frac{h^2}{2}f''(x_0)-\frac{h^3}{6}f'''(x_0)+\frac{h^4}{24}f^{(4)}(x_0)+O(h^5) \tag{2}\\
&f(x_0+h)=f(x_0)+hf'(x_0)+\frac{h^2}{2}f''(x_0)+\frac{h^3}{6}f'''(x_0)+\frac{h^4}{24}f^{(4)}(x_0)+O(h^5) \tag{3}\\
&f(x_0+2h)=f(x_0)+2hf'(x_0)+2h^2f''(x_0)+\frac{8h^3}{6}f'''(x_0)+\frac{16h^4}{24}f^{(4)}(x_0)+O(h^5) \tag{4}
\end{align}

Now, we combine these four equations by taking $(1)-8\times(2)+8\times(3)-(4)$.

\begin{align*}
f(x_0-&2h)-8f(x_0-h)+8f(x_0+2h)-f(x_0+h) \\
&=(1-8+8-1)f(x_0) + (-2+8+8-2)hf'(x_0)+(2-4+4-2)h^2f''(x_0) \\
&~~~~~~~+(-\frac{8}{6}+\frac{8}{6}+\frac{8}{6}-\frac{8}{6})h^3f'''(x_0)+(\frac{16}{24}-\frac{1}{24}+\frac{1}{24}-\frac{16}{24})h^4f^{(4)}(x_0)+O(h^5)\\
&=12hf'(x_0)+O(h^5)
\end{align*}

Rearranging the above equation and solving for $f'(x_0)$ gives,

$$
\boxed{f'(x_0) \approx \frac{1}{12h} \left [f(x_0-2h)-8f(x_0-h)+8f(x_0+h)-f(x_0+2h) \right ]} \text{ with error } O(h^4).
$$

## Richardson's Extrapolation
Richardson's extrapolation is the technique of combining two different numerical approximations 
that depend on a parameter (usually a step size $h$) in order to obtain a new approximation having 
a smaller truncation error. Let $M$ denote some value to be computed, for example,

$$
f'(x) \texttt{ or } f''(x) \texttt{ or } \int_a^b f(x)dx.
$$

Let $N_1(h)$ denote a formula (that depends on a parameter $h$ that can take on different values) 
for computing an approximation to $M$, and suppose that the form of the truncation error of this 
formula is a known infinite series in powers of $h$. For example, the most common case is that the 
truncation error is $O(h^2)$ and is an infinite series with only even powers of $h$, that is,

\begin{equation}
\underbrace{M}_{\substack{\text{exact}\\\text{value}}} = \underbrace{N_1(h)}_{\substack{\text{computed}\\\text{approx.}}} + \underbrace{K_1h^2+K_2h^4+K_3h^6+\dotsm}_\text{turncation error $O(h^2)$} \tag{5}
\end{equation}

where the values $K_i$ are some (possibly unknown) constants.  The parameter $h$ can be any positive 
value, but as $h \rightarrow 0$, the truncation error $\rightarrow 0$; that is, $N_1(h) \rightarrow M$.

Now, consider a new approximation with a different stepsize, say $h/2$,

\begin{equation}
M = N_1\left(\frac{h}{2}\right) + K_1\frac{h^2}{4}+K_2\frac{h^4}{16}+K_3\frac{h^6}{64}+\dotsm \tag{6}
\end{equation}

In order to obtain an $O(h^4)$ approximation to $M$, we compute $4 \times (6) - (5)$, which gives,

\begin{equation}
M = \underbrace{N_1\left(\frac{h}{2}\right)+\frac{N_1\left(\frac{h}{2}\right)-N_1(h)}{3}}_{\text{New approxiamtion $N_2(h)$}} \underbrace{-\frac{K_2}{4}h^4-\frac{5K_3}{16}h^6-\dotsm}_\text{turncation error $O(h^4)$} \tag{7}
\end{equation} 

This is Richardson's Extrapolation - combine two $O(h^2)$ approximations to get a better $O(h^4)$ 
approximation with the simple computation,

\begin{equation*}
N_2(h) = N_1\left(\frac{h}{2}\right)+\frac{N_1\left(\frac{h}{2}\right)-N_1(h)}{3}
\end{equation*}

We can extend this to get an $O(h^6)$ approximation, then an $O(h^8)$ approximation, etc. To do so, 
just keep taking successively smaller stepsizes, but they must \underline{decrease by the same factor}. 
If we continue from equation (7), we can derive a simple formula for all of these approximations. 
Let (7) be rewritten as,

\begin{equation}
M = N_2(h) + K'_2h^4+K'_3h^6+\dotsm \tag{8}
\end{equation}

Then, let $h=\frac{h}{4}$, giving,

\begin{equation}
M = N_1\left(\frac{h}{4}\right) + K_1\frac{h^2}{4^2}+K_2\frac{h^4}{4^4}+K_3\frac{h^6}{4^6}+\dotsm \tag{9}
\end{equation}

Combine $4 \times (9) - (6)$, getting,

\begin{equation}
M = \underbrace{N_1\left(\frac{h}{4}\right)+\frac{N_1\left(\frac{h}{4}\right)-N_1\left(\frac{h}{2}\right)}{3}}_{\text{New approxiamtion $N_2\left(\frac{h}{2}\right)$}} \underbrace{-\frac{K_2}{4}\frac{h^4}{2^4}-\frac{5K_3}{16}\frac{h^6}{2^6}-\dotsm}_\text{turncation error $O(h^4)$} \tag{10}
\end{equation}

Then, combine $16 \times (10) - (8)$, this gives,

\begin{equation}
M = \underbrace{N_2\left(\frac{h}{2}\right)+\frac{N_2\left(\frac{h}{2}\right)-N_2\left(h\right)}{15}}_{\text{New approxiamtion $N_3\left(h\right)$}}+ \underbrace{K''_3h^6+K''_4h^8 + \dotsm}_\text{turncation error $O(h^6)$} \tag{11}
\end{equation} 

**Richardson's Extrapolation Table**

| $O(h^2)$ | $O(h^4)$ | $O(h^6)$ | $O(h^8)$ |
|:------:|:------:|:------:|:------:|
| $N_1(h)$ | | | | 
| $N_1\left(\frac{h}{2}\right)$ | $N_2(h)$ | | | 
| $N_1\left(\frac{h}{4}\right)$ | $N_2\left(\frac{h}{2}\right)$ | $N_3(h)$ | | 
| $N_1\left(\frac{h}{8}\right)$ | $N_2\left(\frac{h}{4}\right)$ | $N_3\left(\frac{h}{2}\right)$ | $N_4(h)$ |

In general, entries are computed by,
$$
N_j\left(\frac{h}{2^i}\right) = N_{j-1}\left(\frac{h}{2^{i+1}}\right) + \frac{N_{j-1}\left(\frac{h}{2^{i+1}}\right) - N_{j-1}\left(\frac{h}{2^i}\right)}{4^{j-1}-1},
$$
where $j \geq 2$ and the truncation error is $O(h^{2j})$.

### Example of Richardson's Extrapolation
Use Richardson's extrapolation and the approximation formula

$$
f'(x_0) \approx \frac{f(x_o + h) - f(x_0 - h)}{2h},
$$

which has a truncation error of the form,

$$
K_1h^2 + K_2h^4 + K_3h^6 + \dotsm,
$$

to approximate $f'(2)=22.1671682$ where $f(x)=xe^x$ and $h=0.2$. Here,

$$
N_1(h)=\frac{f(2+h)-f(2-h)}{2h}=\frac{f(2.2)-f(1.8)}{2(0.2)}=\frac{(2.2)e^{2.2}-(1.8)e^{1.8}}{0.4} =22.4141607
$$

and,

$$
N_1\left(\frac{h}{2}\right)=\frac{f(2+0.1)-f(2-0.1)}{2(0.1)}=\frac{(2.1)e^{2.1}-(1.9)e^{1.9}}{0.2}=22.2287869
$$

then,

$$
N_2(h) = N_1\left(\frac{h}{2}\right)+\frac{N_1\left(\frac{h}{2}\right)-N_1(h)}{3}=22.2287869+\frac{22.2287869-22.4141607}{3}=22.1669956
$$

Next,

$$
N_1\left(\frac{h}{4}\right)=\frac{f(2+0.05)-f(2-0.05)}{2(0.05)}=\frac{(2.05)e^{2.05}-(1.95)e^{1.95}}{0.1}=22.1825649
$$

So,

$$
N_2\left(\frac{h}{2}\right) = N_1\left(\frac{h}{4}\right)+\frac{N_1\left(\frac{h}{4}\right)-N_1\left(\frac{h}{2}\right)}{3}=22.1825649+\frac{22.1825649-22.2287869}{3}=22.1671576
$$

and finally,

$$
N_3(h) = N_2\left(\frac{h}{2}\right)+\frac{N_2\left(\frac{h}{2}\right)-N_2(h)}{15}=22.1671576+\frac{22.1671576-22.1669956}{15}=22.1671684
$$

The first 4 rows and columns of the Richardson's Extrapolation table are as follows:

| $h$ | $N_1$ | $N_2$ | $N_3$ | $N_4$ |
|:-----|:-----|:-----|:-----|:-----|
| $0.2$ | $22.4141607$ | | | | 
| $0.1$ | $22.2287869$ | $22.1669956$ | | | 
| $0.05$ | $22.1825649$ | $22.1671576$ | $22.1671684$ | | 
| $0.025$ | $22.1710169$ | $22.1671676$ | $22.1671683$ | $22.1671683$ |

**Advantages of Richardson's Extrapolation**
Using Richardson's extrapolation has the following advantages:
* obtain high accuracy with little computation
* doesn't require very small values of $h$ to get high accuracy, so roundoff error is not a concern

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.differentiate import derivative

def f(x):
    return 1.25 - 0.25*x - 0.5*x**2 - 0.15*x**3 - 0.1*x**4

# data for plotting a smooth curve
x = np.arange(0, 5.01, 0.01)
y = f(x)

# data for derviatives
x_points = np.arange(0, 5.5, 0.5)
y_points = f(x_points)
print("y =",y_points)

# Forward finite-divided derviatives at each left endpoint
dx = np.diff(x_points)
dy = np.diff(y_points)
fp1x = dy/dx
print("forward =",fp1x)

# Central finite-divided derviatives at each interior point
fp2x = np.gradient(y_points,x_points)
print("central =",fp2x)

# High-order derivative with Richardsons Exptrapolation and interation
res = derivative(f,x_points)
print("iterative =",res.df)

# plot of function and points
plt.scatter(
    x_points,
    y_points,
    color="red",
    label="Highlighted Points",
)
plt.plot(x,y,label='f(x)')
plt.legend()
plt.show()
```
